```python
X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90],
    [20, 30, 40],
    [60, 70, 80]
])

print(X)
print("Shape:", X.shape)
```

```python
mean = np.mean(X, axis=0)

print(mean)
```

```python
result = X[X[:, 0] > 50]

print(result)
```

```python
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

X_standardized = (X - mean) / std

print(X_standardized)
```
